Pengenalan Library Phyton

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # For image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

Membaca dan menampilkan image

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://i.pinimg.com/736x/c0/4f/c6/c04fc6ae79dec526abcd17c3f57db493.jpg"]

for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

Melihat ukuran file image

```python
tinggi = image_2.shape[0]
lebar  = image_2.shape[1]

print("Resolusi image: tinggi x lebar = ", tinggi, "x", lebar)
cv2_imshow(image_2)
```

Memberi garis horizontal dengan mengakses pixel

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(lebar):
  image_3[int((tinggi)/2), y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

TUGAS PRAKTIKUM D2

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_4 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat garis vertikal
for x in range(tinggi):
  image_3[x, int((lebar)/2)] = [255, 255, 255]

# Membuat garis diagonal
for y in range(tinggi):
    x = int(y * (lebar - 1) / (tinggi - 1))
    image_4[y, x] = [255, 255, 255]

for y in range(tinggi):
  x = int((tinggi - 1 - y) * (lebar - 1) / (tinggi - 1))
  image_4[y, x] = [255, 255, 255]

print("1. Membuat garis vertikal dan diagonal pada citra")
final_frame = cv.hconcat((image_2, image_3, image_4))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tinggi, lebar = image_3.shape[:2]
garis = 156

y = int(tinggi/2)

x_awal  = int((lebar - garis) / 2)
x_akhir = x_awal - garis

image_3[y, x_awal:x_akhir] = [255, 255, 255]

print("2. Membuat garis horizontal dengan panjang 100")
final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
