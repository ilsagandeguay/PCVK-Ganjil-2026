2. Implementasikan transformasi contrast pada Google Colaboratory menggunakan formula yang terdapat pada bagian Ulasan Teori untuk kontras, sehingga menghasilkan keluaran seperti berikut:

```python
print(' Mengubah tingkat kecerahan dan kontras citra ')
print(' -------------------------------------------- ')

try:
    brightness = int(input('Masukkan nilai kecerahan: '))
    contrast = float(input('Masukkan nilai kontras: '))

except ValueError:
    print('Masukkan nilai yang benar')

# mengambil file dari drive
original = cv.imread('/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/sunset.jpg')
original_contrast = io.imread('https://i.pinimg.com/736x/b5/21/6a/b5216a303633d77df76ec1aaa6752efa.jpg')

contrast_img2 =cv.cvtColor(original_contrast, cv.COLOR_BGR2RGB)

# membuat citra untuk hasil kontras
contrast_image = np.zeros(original.shape, original.dtype)
contrast_image2 = np.zeros(contrast_img2.shape, contrast_img2.dtype)

# akses setiap pixel citra 1
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):

            # Mengubah kontras
            contrast_image[y,x,c] = np.clip(
                contrast * (original[y,x,c]) + brightness, 0, 255
            )

# akses setiap pixel citra 2
for y in range(contrast_img2.shape[0]):
    for x in range(contrast_img2.shape[1]):
        for c in range(contrast_img2.shape[2]):

            # Mengubah kontras
            contrast_image2[y,x,c] = np.clip(
                contrast * (contrast_img2[y,x,c]) + brightness, 0, 255
            )

# menampilkan citra asli dan hasil kontras
final_contrast = cv.hconcat([original, contrast_image])
final_contrast2 = cv.hconcat([contrast_img2, contrast_image2])

print('Citra asli dan hasil perubahan kontras:')
cv2_imshow(final_contrast)

print('Citra asli dan hasil perubahan kontras:')
cv2_imshow(final_contrast2)
```

3. Implementasikan transformasi logarithmic brightness pada Google Colaboratory menggunakan formula yang terdapat pada bagian Ulasan Teori untuk transformasi log, sehingga menghasilkan keluaran seperti berikut:

```python
print(' Transformasi Logarithmic Brightness ')
print(' ---------------------------------- ')

# memasukkan nilai kecerahan
try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Masukkan nilai yang benar')

# mengambil file dari drive
original = cv.imread('/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/houses.jpg')

# mengambil citra kedua dari internet
original_log = io.imread('https://i.pinimg.com/736x/b5/21/6a/b5216a303633d77df76ec1aaa6752efa.jpg')

# io.imread menghasilkan RGB, jadi tidak perlu cvtColor
log_img2 = cv.cvtColor(original_log, cv.COLOR_BGR2RGB)


# menentukan nilai konstanta c
c = 255 / np.log(256)

# membuat citra hasil transformasi
log_image = np.zeros(original.shape, original.dtype)
log_image2 = np.zeros(log_img2.shape, log_img2.dtype)

# ==============================
# CITRA 1
# ==============================

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for ch in range(original.shape[2]):

            # Rumus logarithmic + brightness
            log_image[y,x,ch] = np.clip(
                c * np.log(1 + original[y,x,ch]) + brightness,
                0,
                255
            )


# ==============================
# CITRA 2
# ==============================

for y in range(log_img2.shape[0]):
    for x in range(log_img2.shape[1]):
        for ch in range(log_img2.shape[2]):

            # Rumus logarithmic + brightness
            log_image2[y,x,ch] = np.clip(
                c * np.log(1 + log_img2[y,x,ch]) + brightness,
                0,
                255
            )


# Menggabungkan citra asli dan hasil transformasi
final_frame1 = cv.hconcat([original, log_image])
final_frame2 = cv.hconcat([log_img2, log_image2])

# Menampilkan hasil
print('Citra 1:')
cv2_imshow(final_frame1)

print('Citra 2:')
cv2_imshow(final_frame2)
```

4. Implementasikan transformasi grayscale menggunakan metode averaging, lightness, dan luminance pada Google Colaboratory menggunakan formula yang terdapat pada bagian Ulasan Teori

```python
print(' Transformasi Grayscale ')
print(' ---------------------- ')

# ==========================================
# MENGAMBIL 2 CITRA
# ==========================================

# Gambar pertama dari Drive
original = cv.imread(
    '/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/peppers.jpg'
)

# Gambar kedua dari URL
original2 = io.imread(
    'https://i.pinimg.com/736x/b5/21/6a/b5216a303633d77df76ec1aaa6752efa.jpg'
)

# Mengubah gambar pertama dari BGR ke RGB
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

# Menyesuaikan ukuran gambar kedua dengan gambar pertama
original2 = cv.resize(
    original2,
    (original_rgb.shape[1], original_rgb.shape[0])
)

# ==========================================
# MEMBUAT CITRA GRAYSCALE
# ==========================================

# Gambar 1
grayscale_averaging = np.zeros(
    original_rgb.shape[:2], dtype=np.uint8
)

grayscale_lightness = np.zeros(
    original_rgb.shape[:2], dtype=np.uint8
)

grayscale_luminance = np.zeros(
    original_rgb.shape[:2], dtype=np.uint8
)

# Gambar 2
grayscale_averaging2 = np.zeros(
    original2.shape[:2], dtype=np.uint8
)

grayscale_lightness2 = np.zeros(
    original2.shape[:2], dtype=np.uint8
)

grayscale_luminance2 = np.zeros(
    original2.shape[:2], dtype=np.uint8
)

# ==========================================
# AKSES SETIAP PIXEL GAMBAR 1
# ==========================================

for y in range(original_rgb.shape[0]):

    for x in range(original_rgb.shape[1]):

        R = original_rgb[y, x, 0]
        G = original_rgb[y, x, 1]
        B = original_rgb[y, x, 2]

        # Metode Averaging
        grayscale_averaging[y, x] = np.clip(
            (R + G + B) / 3,
            0,
            255
        )

        # Metode Lightness
        grayscale_lightness[y, x] = np.clip(
            (max(R, G, B) + min(R, G, B)) / 2,
            0,
            255
        )

        # Metode Luminance
        grayscale_luminance[y, x] = np.clip(
            0.21 * R + 0.72 * G + 0.07 * B,
            0,
            255
        )

# ==========================================
# AKSES SETIAP PIXEL GAMBAR 2
# ==========================================

for y in range(original2.shape[0]):

    for x in range(original2.shape[1]):

        R = original2[y, x, 0]
        G = original2[y, x, 1]
        B = original2[y, x, 2]

        # Metode Averaging
        grayscale_averaging2[y, x] = np.clip(
            (R + G + B) / 3,
            0,
            255
        )

        # Metode Lightness
        grayscale_lightness2[y, x] = np.clip(
            (max(R, G, B) + min(R, G, B)) / 2,
            0,
            255
        )

        # Metode Luminance
        grayscale_luminance2[y, x] = np.clip(
            0.21 * R + 0.72 * G + 0.07 * B,
            0,
            255
        )

# ==========================================
# MENAMPILKAN HASIL DENGAN PLT
# ==========================================

plt.figure(figsize=(16, 8))

# ------------------------------------------
# GAMBAR 1
# ------------------------------------------

plt.subplot(2, 4, 1)
plt.imshow(original_rgb)
plt.title('Gambar 1 - Asli')
plt.axis('on')

plt.subplot(2, 4, 2)
plt.imshow(grayscale_averaging, cmap='gray')
plt.title('Gambar 1 - Averaging')
plt.axis('on')

plt.subplot(2, 4, 3)
plt.imshow(grayscale_lightness, cmap='gray')
plt.title('Gambar 1 - Lightness')
plt.axis('on')

plt.subplot(2, 4, 4)
plt.imshow(grayscale_luminance, cmap='gray')
plt.title('Gambar 1 - Luminance')
plt.axis('on')

# ------------------------------------------
# GAMBAR 2
# ------------------------------------------

plt.subplot(2, 4, 5)
plt.imshow(original2)
plt.title('Gambar 2 - Asli')
plt.axis('on')

plt.subplot(2, 4, 6)
plt.imshow(grayscale_averaging2, cmap='gray')
plt.title('Gambar 2 - Averaging')
plt.axis('on')

plt.subplot(2, 4, 7)
plt.imshow(grayscale_lightness2, cmap='gray')
plt.title('Gambar 2 - Lightness')
plt.axis('on')

plt.subplot(2, 4, 8)
plt.imshow(grayscale_luminance2, cmap='gray')
plt.title('Gambar 2 - Luminance')
plt.axis('on')

plt.tight_layout()
plt.show()
```

6. Buat Gamma Correction sesuai dengan petunjuk berikut

    Percobaan ini akan meminta anda membuat Gamma Correction. Pada percobaan ini, nilai Gamma akan diset dengan meminta masukan dari pengguna. Berikut adalah kode untuk meminta masukan nilai dari pengguna. Lanjutkan kode tersebut dengan membuat image dengan gamma correction sesuai rumus yang telah diberikan.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from skimage import io

print(' Gamma Correction pada citra ')
print('----------------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')

# =========================
# GAMBAR 1
# =========================
original1 = cv.imread(
    '/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/houses.jpg'
)

# Membuat citra hasil
gamma_image1 = np.zeros_like(original1)

# Gamma Correction gambar 1
for y in range(original1.shape[0]):
    for x in range(original1.shape[1]):
        for c in range(original1.shape[2]):
            gamma_image1[y, x, c] = np.clip(
                255 * ((original1[y, x, c] / 255) ** gamma),
                0,
                255
            )

# Konversi BGR ke RGB
original1_rgb = cv.cvtColor(original1, cv.COLOR_BGR2RGB)
gamma_image1_rgb = cv.cvtColor(gamma_image1, cv.COLOR_BGR2RGB)


# =========================
# GAMBAR 2
# =========================
original2 = io.imread(
    'https://i.pinimg.com/736x/b5/21/6a/b5216a303633d77df76ec1aaa6752efa.jpg'
)

# Jika gambar memiliki channel alpha (RGBA)
if original2.shape[2] == 4:
    original2 = original2[:, :, :3]

# Membuat citra hasil
gamma_image2 = np.zeros_like(original2)

# Gamma Correction gambar 2
for y in range(original2.shape[0]):
    for x in range(original2.shape[1]):
        for c in range(original2.shape[2]):
            gamma_image2[y, x, c] = np.clip(
                255 * ((original2[y, x, c] / 255) ** gamma),
                0,
                255
            )


# =========================
# MENAMPILKAN HASIL
# =========================
plt.figure(figsize=(12, 8))

# Gambar 1 original
plt.subplot(2, 2, 1)
plt.imshow(original1_rgb)
plt.title('Gambar 1 - Original')
plt.xlabel('X')
plt.ylabel('Y')

# Gambar 1 hasil Gamma
plt.subplot(2, 2, 2)
plt.imshow(gamma_image1_rgb)
plt.title('Gambar 1 - Gamma Correction (γ = ' + str(gamma) + ')')
plt.xlabel('X')
plt.ylabel('Y')

# Gambar 2 original
plt.subplot(2, 2, 3)
plt.imshow(original2)
plt.title('Gambar 2 - Original')
plt.xlabel('X')
plt.ylabel('Y')

# Gambar 2 hasil Gamma
plt.subplot(2, 2, 4)
plt.imshow(gamma_image2)
plt.title('Gambar 2 - Gamma Correction (γ = ' + str(gamma) + ')')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()
```

7. Buat Simulasi Image Depth

    Percobaan ini digunakan sebagai simulasi dari proses kuantisasi citra. Pada kuantisasi citra, pixel dapat direpresentasikan dengan n-bit kedalaman (default menggunakan 8-bit). Pada pixel 8-bit, warna yang memungkinkan adalah 256 warna, dari 0 (0000 0000) hingga 255(1111 1111). Pada pixel 7-bit, warna yang memungkinkan adalah 128 warna, dari 0 (000 0000) hingga 127 (111 1111). Kemungkinan warna didapat dari pangkat 2 jumlah bit. Jika 7bit, maka jumlah warnanya adalah 27 = 128, dst.

```python
print(' Simulasi Image Depth ')
print('----------------------')

try:
    bit_depth = int(input('Masukkan nilai bit depth (1-8): '))

    if bit_depth < 1 or bit_depth > 8:
        print('Bit depth harus antara 1 sampai 8')
    else:
        # Menghitung level
        level = 255 / (pow(2, bit_depth) - 1)

        # Membaca citra grayscale
        original = cv.imread(
            '/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/peppers.jpg',
            cv.IMREAD_GRAYSCALE
        )

        # Membuat citra hasil
        depth_image = np.zeros(original.shape, original.dtype)

        # Proses kuantisasi
        for y in range(original.shape[0]):
            for x in range(original.shape[1]):
                depth_image[y, x] = np.round(
                    original[y, x] / level
                ) * level

        # Menampilkan hasil
        plt.figure(figsize=(12, 5))

        plt.subplot(1, 2, 1)
        plt.imshow(original, cmap='gray')
        plt.title('Citra Original (8-bit)')
        plt.xlabel('X')
        plt.ylabel('Y')

        plt.subplot(1, 2, 2)
        plt.imshow(depth_image, cmap='gray')
        plt.title(
            str(bit_depth) + '-bit (' +
            str(pow(2, bit_depth)) + ' level)'
        )
        plt.xlabel('X')
        plt.ylabel('Y')

        plt.tight_layout()
        plt.show()
except ValueError:
    print('Input yang dimasukkan tidak valid. Harap masukkan angka.')
except Exception as e:
    print(f"Terjadi kesalahan: {e}")
```

8. Buat modul average denoising sesuai dengan rumus yang telah diberikan pada sub bab sebelumnya. Citra asli sudah disediakan pada /images/galaxy.jpg. 100 Citra dengan Gaussian Noise sudah disediakan pada /images/noises/*.jpg Anda dapat menggunakan code berikut untuk membaca semua image dalam satu folder , gunakan modul glob (import glob).

10 Citra Noises

```python
# ==========================================
# MEMBACA CITRA ASLI
# ==========================================

original = cv.imread(
    '/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/galaxy.jpg'
)

# Konversi BGR ke RGB untuk matplotlib
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)


# ==========================================
# MEMBACA 10 CITRA GAUSSIAN NOISE
# ==========================================

images = []

for i in range(1, 11):
    file = f'/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/noises/{i}.jpg'

    img = cv.imread(file)

    if img is not None:
        # Konversi BGR ke RGB
        img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
        images.append(img_rgb)

print('Jumlah citra noise:', len(images))


# ==========================================
# MODUL AVERAGE DENOISING
# ==========================================

def average_denoising(images):

    images = np.array(images, dtype=np.float32)

    average_image = np.mean(images, axis=0)

    average_image = np.clip(
        average_image, 0, 255
    ).astype(np.uint8)

    return average_image


# Menjalankan modul
average_image = average_denoising(images)


# ==========================================
# MENGHITUNG PSNR
# ==========================================

mse = np.mean(
    (original_rgb.astype(np.float32) -
     average_image.astype(np.float32)) ** 2
)

if mse == 0:
    psnr = float('inf')
else:
    psnr = 10 * np.log10((255 ** 2) / mse)

print('MSE  :', mse)
print('PSNR :', psnr, 'dB')


# ==========================================
# MENAMPILKAN HASIL
# ==========================================

plt.figure(figsize=(15, 5))

# Citra asli
plt.subplot(1, 3, 1)
plt.imshow(original_rgb)
plt.title('Citra Asli')
plt.xlabel('X')
plt.ylabel('Y')

# Gaussian Noise
plt.subplot(1, 3, 2)
plt.imshow(images[0])
plt.title('Gaussian Noise 1')
plt.xlabel('X')
plt.ylabel('Y')

# Average Denoising
plt.subplot(1, 3, 3)
plt.imshow(average_image)
plt.title('Average Denoising 10')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()

print('PSNR Average Denoising = {:.2f} dB'.format(psnr))
```

9. Buat Image Making
    Buat image masking untuk image berikut. Image kiri adalah image asli (images/couple.tiff),
    sedangkan image paling kanan adalah hasilnya:
    Lakukan percobaan menggunakan operator lain dan tunjukkan hasilnya pada modul ini.
    Tuliskan hasil analisa anda kenapa citra keluarannya seperti itu.


```python
# Membaca citra
original = cv.imread(
    '/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/couple.tiff'
)

# Konversi BGR ke RGB
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

# Ukuran citra
height, width = original.shape[:2]

# ==========================================
# MEMBUAT MASK DUA LINGKARAN
# ==========================================

mask = np.zeros((height, width), dtype=np.uint8)

# Lingkaran orang sebelah kiri
cv.circle(
    mask,
    (int(width * 0.30), int(height * 0.25)),
    int(height * 0.20),
    255,
    -1
)

# Lingkaran orang sebelah kanan
cv.circle(
    mask,
    (int(width * 0.70), int(height * 0.25)),
    int(height * 0.20),
    255,
    -1
)

# Ubah mask menjadi 3 channel
mask_3ch = cv.cvtColor(mask, cv.COLOR_GRAY2BGR)

# ==========================================
# OPERATOR AND
# ==========================================

hasil_and = cv.bitwise_and(original, mask_3ch)

# ==========================================
# OPERATOR OR
# ==========================================

hasil_or = cv.bitwise_or(original, mask_3ch)

# ==========================================
# OPERATOR NAND
# NAND = NOT(AND)
# ==========================================

hasil_and_temp = cv.bitwise_and(original, mask_3ch)
hasil_nand = cv.bitwise_not(hasil_and_temp)

# ==========================================
# OPERATOR XOR
# ==========================================

hasil_xor = cv.bitwise_xor(original, mask_3ch)

# ==========================================
# OPERATOR NOT
# ==========================================

hasil_not = cv.bitwise_not(original)

# ==========================================
# KONVERSI HASIL KE RGB
# ==========================================

hasil_and_rgb = cv.cvtColor(hasil_and, cv.COLOR_BGR2RGB)
hasil_or_rgb = cv.cvtColor(hasil_or, cv.COLOR_BGR2RGB)
hasil_nand_rgb = cv.cvtColor(hasil_nand, cv.COLOR_BGR2RGB)
hasil_xor_rgb = cv.cvtColor(hasil_xor, cv.COLOR_BGR2RGB)
hasil_not_rgb = cv.cvtColor(hasil_not, cv.COLOR_BGR2RGB)

# ==========================================
# MENAMPILKAN HASIL
# ==========================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(original_rgb)
plt.title('Citra Asli')
plt.xlabel('X')
plt.ylabel('Y')

plt.subplot(2, 3, 2)
plt.imshow(mask, cmap='gray')
plt.title('Mask Dua Lingkaran')
plt.xlabel('X')
plt.ylabel('Y')

plt.subplot(2, 3, 3)
plt.imshow(hasil_and_rgb)
plt.title('Operator AND')
plt.xlabel('X')
plt.ylabel('Y')

plt.subplot(2, 3, 4)
plt.imshow(hasil_or_rgb)
plt.title('Operator OR')
plt.xlabel('X')
plt.ylabel('Y')

plt.subplot(2, 3, 5)
plt.imshow(hasil_nand_rgb)
plt.title('Operator NAND')
plt.xlabel('X')
plt.ylabel('Y')

plt.subplot(2, 3, 6)
plt.imshow(hasil_xor_rgb)
plt.title('Operator XOR')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()

# Menampilkan NOT secara terpisah
plt.figure(figsize=(6, 5))

plt.imshow(hasil_not_rgb)
plt.title('Operator NOT')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()
```

10. Carilah foto malam hari yang gelap. Namun, beberapa bagian gambar memiliki detail yang sebenarnya cukup baik. Pilih salah satu metode berikut dan implementasikan pada foto tersebut (Linear Brightness, Contrast, Logarithmic Brightness, Gamma Correction). Sebutkan alasan dan resiko metode yang Anda gunakan.

```python
# Membaca foto Milky Way
original = cv.imread('/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Week3/img.jpg')

# Konversi BGR ke RGB
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

# Nilai gamma
gamma = 0.5

# Membuat citra hasil
gamma_image = np.zeros_like(original)

# Proses Gamma Correction
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            gamma_image[y, x, c] = np.clip(
                255 * ((original[y, x, c] / 255) ** gamma),
                0,
                255
            )

# Konversi hasil ke RGB
gamma_rgb = cv.cvtColor(gamma_image, cv.COLOR_BGR2RGB)

# Menampilkan hasil
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title('Foto Milky Way Asli')
plt.xlabel('X')
plt.ylabel('Y')

plt.subplot(1, 2, 2)
plt.imshow(gamma_rgb)
plt.title('Gamma Correction (γ = 0.5)')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()
```

11. Buka gambar crayfish.jpg pada folder Images. Perbaiki kualitas pada citra tersebut dengan metode operasi dasar citra apapun. Diperbolehkan menggunakan lebih dari 1 metode. Kemudian jelaskan:
Metode yang dipilih

*   Metode yang dipilih
*   Alasan Pemilihan
*   Menentukan paremater terbaik
*   Menghasilkan citra dengan menunjukkan before-after.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# 1. MEMBACA CITRA
# ==========================================================

path = '/content/drive/MyDrive/Politeknik Negeri Malang/Matakuliah/Pengenalan Citra dan Visi Komputer/Images/crayfish.jpg'

original = cv.imread(path)

if original is None:
    print('Citra tidak ditemukan. Periksa kembali path file.')
else:
    print('Citra berhasil dibaca.')

# Konversi BGR ke RGB
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)


# ==========================================================
# 2. MENAMPILKAN CITRA ASLI
# ==========================================================

plt.figure(figsize=(7, 5))

plt.imshow(original_rgb)
plt.title('Citra Asli - Crayfish')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()


# ==========================================================
# 3. INPUT NILAI KECERAHAN
# ==========================================================

print('==========================================')
print('       LINEAR BRIGHTNESS CITRA')
print('==========================================')

try:
    brightness = int(
        input('Masukkan nilai kecerahan (-255 sampai 255): ')
    )
except ValueError:
    print('Masukkan nilai yang benar.')
    brightness = 50


# ==========================================================
# 4. PROSES LINEAR BRIGHTNESS
# ==========================================================

brightness_image = np.zeros_like(original)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):

            brightness_image[y, x, c] = np.clip(
                original[y, x, c] + brightness,
                0,
                255
            )


# ==========================================================
# 5. KONVERSI HASIL KE RGB
# ==========================================================

brightness_rgb = cv.cvtColor(
    brightness_image,
    cv.COLOR_BGR2RGB
)


# ==========================================================
# 6. BEFORE - AFTER
# ==========================================================

plt.figure(figsize=(14, 6))

# Before
plt.subplot(1, 2, 1)

plt.imshow(original_rgb)

plt.title('Before - Citra Asli')

plt.xlabel('X')
plt.ylabel('Y')


# After
plt.subplot(1, 2, 2)

plt.imshow(brightness_rgb)

plt.title(
    f'After - Linear Brightness (+{brightness})'
)

plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()
```
